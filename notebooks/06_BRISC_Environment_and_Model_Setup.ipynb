```python
import torch

device = torch.device("cuda")

x = torch.randn(1024, 1024, device=device)
y = torch.matmul(x, x)

torch.cuda.synchronize()

print("Device:", y.device)
print("GPU computation successful!")
print("Result shape:", y.shape)
```
